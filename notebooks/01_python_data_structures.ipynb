## set

```python
sample_set = {1, 5, 8, 9, 4, 5, 6, "rajat", 6.1}    # duplicates (5) are kept only once
sample_set
```

```python
sample_set.remove(1)
sample_set
```

```python
sample_set
```

```python
sample_set.remove(6)
sample_set
```

```python
sample_set
```

```python
sample_set.add(152)
sample_set
```

```python
sample_set
```

```python
sample_set.add(12)
sample_set
```

```python
set_as_tuple = tuple(sample_set)    # convert the set to a tuple
set_as_tuple
```

```python
# Tuples are immutable, so they have no add() method
try:
    set_as_tuple.add(12)
except AttributeError as err:
    print("AttributeError:", err)
```

```python
back_to_set = set(set_as_tuple)    # and back to a set
back_to_set
```
